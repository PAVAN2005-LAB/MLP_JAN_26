```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np

# Load dataset
df = pd.read_csv('GA_5_dataset.csv')

# Step 1: Separate X and y
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

# Step 2: Split dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Questions 1 & 2: Linear Regression
lr = LinearRegression(fit_intercept=False)
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
r2_lr = r2_score(y_test, y_pred_lr)
max_abs_coef_idx = np.argmax(np.abs(lr.coef_))

# Questions 3 & 4: Ridge Regression
ridge = Ridge(solver='sag', tol=0.0005, random_state=42)
ridge.fit(X_train, y_train)
y_pred_ridge = ridge.predict(X_test)
r2_ridge = r2_score(y_test, y_pred_ridge)
intercept_ridge = ridge.intercept_

# Questions 5 & 6: Lasso Regression
lasso = Lasso(alpha=100, random_state=42)
lasso.fit(X_train, y_train)
y_pred_lasso = lasso.predict(X_test)
r2_lasso = r2_score(y_test, y_pred_lasso)
coef_in_range = np.sum((lasso.coef_ >= -1) & (lasso.coef_ <= 1))

# Question 7: KNeighborsRegressor
knn = KNeighborsRegressor(n_neighbors=10, p=1)
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)
rmse_knn = np.sqrt(mean_squared_error(y_test, y_pred_knn))

# Question 8: DecisionTreeRegressor
dt = DecisionTreeRegressor(max_depth=10, min_samples_split=6, min_samples_leaf=6, random_state=42)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)
rmse_dt = np.sqrt(mean_squared_error(y_test, y_pred_dt))

# Questions 9 & 10: AdaBoostRegressor with GridSearchCV
param_grid = {
    'n_estimators': [10, 50, 100, 200, 500],
    'learning_rate': [0.1, 0.5, 1, 2]
}
ada = AdaBoostRegressor(random_state=42)
grid_search = GridSearchCV(ada, param_grid, cv=4)
grid_search.fit(X_train, y_train)
best_ada = grid_search.best_estimator_
y_pred_ada = best_ada.predict(X_test)
r2_ada = r2_score(y_test, y_pred_ada)
best_n_estimators = grid_search.best_params_['n_estimators']

# Results
print(f"Q1: Linear Regression R2 score: {r2_lr:.3f}")
print(f"Q2: Index of highest absolute coefficient: {max_abs_coef_idx}")
print(f"Q3: Ridge R2 score: {r2_ridge:.3f}")
print(f"Q4: Ridge intercept: {intercept_ridge:.3f}")
print(f"Q5: Lasso R2 score: {r2_lasso:.3f}")
print(f"Q6: Count of coefficients in range [-1, 1]: {coef_in_range}")
print(f"Q7: KNN RMSE: {rmse_knn:.3f}")
print(f"Q8: DT RMSE: {rmse_dt:.3f}")
print(f"Q9: AdaBoost R2 score: {r2_ada:.3f}")
print(f"Q10: Best n_estimators: {best_n_estimators}")
```
