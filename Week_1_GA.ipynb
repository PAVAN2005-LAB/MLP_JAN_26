

q.1 How many unknown ("?) values are present in the dataset? Remove/Delete unknown ("?) values present in the dataset to make it null value.
Remove/Delete - means it will show NAN in place of "?"
Note: If there is no value present in the dataset it is represented as NAN(read pandas documentation for all the other ways to represent null values) in data



```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('Week1_GA_dataset.csv')

# Count unknown values ('?') before replacement
unknown_count_before = (df == '?').sum().sum()
print(f"Number of '?' values before replacement: {unknown_count_before}")

# Replace '?' with NaN
df_replaced = df.replace('?', np.nan)

# Count NaN values after replacement to confirm
nan_count_after = df_replaced.isna().sum().sum()
print(f"Number of NaN values after replacement (should be equal to original '?' count): {nan_count_after}")

# Display the first few rows of the DataFrame with replaced values to show the effect
print("\nDataFrame head after replacing '?' with NaN:")
print(df_replaced.head())
```

```python
# Calculate the number of missing values per row
missing_values_per_row = df_replaced.isnull().sum(axis=1)

# Filter out rows where the number of missing values is strictly greater than 2
df_filtered = df_replaced[missing_values_per_row <= 2]

# Get the number of remaining samples (rows)
remaining_samples = df_filtered.shape[0]

print(f"Number of samples remaining after dropping rows with more than 2 missing values: {remaining_samples}")

# Display the shape of the filtered DataFrame to confirm
print(f"Shape of the DataFrame after filtering: {df_filtered.shape}")
```

```python
# Get all unique years from the 'Year' column
unique_years = df_replaced['Year'].unique()

# Sort years in descending order to get the most recent ones first
sorted_years = np.sort(unique_years)[::-1]

# Select the six most recent years
six_most_recent_years = sorted_years[:6]

# Filter the DataFrame to include only samples from these six years
df_recent_years = df_replaced[df_replaced['Year'].isin(six_most_recent_years)]

# Count the number of samples (rows) in the filtered DataFrame
num_samples_recent_years = df_recent_years.shape[0]

print(f"The six most recent years are: {six_most_recent_years.tolist()}")
print(f"Number of samples belonging to the six most recent years: {num_samples_recent_years}")
```

```python
# Filter the DataFrame for rows where num_rooms is 3 and num_bathrooms is 3
filtered_samples = df_replaced[(df_replaced['num_rooms'] == 3) & (df_replaced['num_bathrooms'] == 3)]

# Count the number of rows in the filtered DataFrame
num_matching_samples = filtered_samples.shape[0]

print(f"Number of samples with num_rooms = 3 and num_bathrooms = 3: {num_matching_samples}")
```

```python
# Filter the DataFrame for rows where num_rooms is 3 OR num_bathrooms is 3
filtered_samples_or = df_replaced[(df_replaced['num_rooms'] == 3) | (df_replaced['num_bathrooms'] == 3)]

# Count the number of rows in the filtered DataFrame
num_matching_samples_or = filtered_samples_or.shape[0]

print(f"Number of samples with num_rooms = 3 OR num_bathrooms = 3: {num_matching_samples_or}")
```

```python
# Filter for Year == 2022
df_2022 = df_replaced[df_replaced['Year'] == 2022]

# Filter for Locality == 'Greenwich'
df_greenwich = df_2022[df_2022['Locality'] == 'Greenwich']

# Filter for num_rooms == 3
df_rooms_3 = df_greenwich[df_greenwich['num_rooms'] == 3]

# Filter for Face == 'North' or Face == 'East'
df_final = df_rooms_3[(df_rooms_3['Face'] == 'North') | (df_rooms_3['Face'] == 'East')]

# Count the number of samples in the final filtered DataFrame
num_houses = df_final.shape[0]

print(f"Number of houses in Year 2022, located in Greenwich, with num_rooms == 3 and facing North or East: {num_houses}")
```

```python
# Ensure 'Date' column is datetime type
df_replaced['Date'] = pd.to_datetime(df_replaced['Date'])

# Filter for samples in August (month 8)
df_august = df_replaced[df_replaced['Date'].dt.month == 8]

# Count the number of samples
num_samples_august = df_august.shape[0]

print(f"Number of samples corresponding to the month of August: {num_samples_august}")
```

```python
# Calculate the average 'Sale Price' for each 'Locality'
average_sale_price_by_locality = df_replaced.groupby('Locality')['Sale Price'].mean()

# Find the Locality with the highest average 'Sale Price'
highest_avg_sale_price_locality = average_sale_price_by_locality.idxmax()
max_avg_sale_price = average_sale_price_by_locality.max()

print(f"The Locality with the highest average Sale Price is '{highest_avg_sale_price_locality}' with an average of ${max_avg_sale_price:,.2f}.")
```
