```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier, VotingClassifier, BaggingClassifier, RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score
import numpy as np

# Load dataset
wine = load_wine()
X, y = wine.data, wine.target

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

# Task 1: Logistic Regression
log_reg = LogisticRegression(max_iter=10000, C=0.5, random_state=1, solver='sag')
log_reg.fit(X_train, y_train)
y_pred_log = log_reg.predict(X_test)
acc_log = accuracy_score(y_test, y_pred_log)
n_iter_log = log_reg.n_iter_[0] # n_iter_ returns an array for each class in multi-class

# Task 2: Decision Tree Hyperparameter Tuning
dt_params = {
    'criterion': ['entropy', 'gini'],
    'splitter': ['random', 'best'],
    'min_samples_leaf': [2, 4, 6, 8, 10],
    'max_depth': [3, 4, 5, 6]
}
dt = DecisionTreeClassifier(random_state=1)
grid_dt = GridSearchCV(dt, dt_params, cv=4)
grid_dt.fit(X_train, y_train)
best_dt = grid_dt.best_estimator_
acc_dt = accuracy_score(y_test, best_dt.predict(X_test))
best_max_depth = grid_dt.best_params_['max_depth']

# Task 3: AdaBoost Hyperparameter Tuning
ada_params = {
    'n_estimators': [100, 500, 1000],
    'learning_rate': [0.5, 1, 2]
}
ada = AdaBoostClassifier(random_state=1)
grid_ada = GridSearchCV(ada, ada_params, cv=4)
grid_ada.fit(X_train, y_train)
best_ada = grid_ada.best_estimator_
acc_ada = accuracy_score(y_test, best_ada.predict(X_test))
best_n_estimators = grid_ada.best_params_['n_estimators']

# Task 4: Voting Classifier
bag = BaggingClassifier(random_state=1)
rf = RandomForestClassifier(random_state=1)
gb = GradientBoostingClassifier(random_state=1)

voting_clf = VotingClassifier(estimators=[
    ('bag', bag),
    ('rf', rf),
    ('gb', gb)
])
voting_clf.fit(X_train, y_train)
acc_voting = accuracy_score(y_test, voting_clf.predict(X_test))

print(f"Task 1: LogReg Acc: {acc_log:.3f}, Iter: {n_iter_log}")
print(f"Task 2: DT Acc: {acc_dt:.3f}, Best max_depth: {best_max_depth}")
print(f"Task 3: AdaBoost Acc: {acc_ada:.3f}, Best n_estimators: {best_n_estimators}")
print(f"Task 4: Voting Acc: {acc_voting:.3f}")
```
